## Phase 1 & 2: Parcel Prioritization (Merge Sort) & Capacity Optimization (0/1 Knapsack DP)

```python
parcels = load_parcels('datasets/parcels_sample.csv')
sorted_parcels = merge_sort_parcels(parcels)

print('=== Sorted Parcels (Earliest Deadline & Highest Priority) ===')
df_sorted = pd.DataFrame([vars(p) for p in sorted_parcels])
display(df_sorted)

VAN_CAPACITY = 25
max_val, selected_parcels, dp_table = optimize_van_loading_dp(sorted_parcels, VAN_CAPACITY)

print(f'\n=== Optimal Packing Plan (Capacity: {VAN_CAPACITY} kg | Max Priority Score: {max_val}) ===')
df_selected = pd.DataFrame([vars(p) for p in selected_parcels])
display(df_selected)
```

## Phase 3: Graph Representation, Dijkstra's Algorithm & Nearest-Neighbor TSP

```python
DEPOT = 'Central_Hub'
graph = load_graph('datasets/delivery_graph.csv')
dists, paths = dijkstra_shortest_paths(graph, DEPOT)

targets = list(dict.fromkeys(p.destination for p in selected_parcels))
visit_order, detailed_route, total_km = tsp_nearest_neighbor(graph, DEPOT, targets)

print('Stop Sequence     :', ' -> '.join(visit_order))
print('Full Road Traverse:', ' -> '.join(detailed_route))
print(f'Total Route Length: {total_km:.2f} km')
```
